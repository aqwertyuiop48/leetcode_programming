```python
!pip install pyforces-cli
import json
import os
import subprocess

choice = os.environ.get("CHOICE", "").strip()
if choice not in {str(index) for index in range(10)}:
    raise ValueError("Set CHOICE to a valid pyforces option (0-9) before running the notebook")

answers = [choice]

if choice == "0":
    header = os.environ.get("PYFORCES_HEADER_JSON", "").strip()
    if not header:
        raise ValueError("Set the PYFORCES_HEADER_JSON Codespaces secret before choosing 0")
    json.loads(header)  # Check that it is valid JSON before starting the CLI
    answers.append(header)

result = subprocess.run(
    ["pyforces", "config"],
    input="\n".join(answers) + "\n",
    text=True,
    capture_output=True,
)

print(result.stdout, end="")
if result.stderr:
    print(result.stderr, end="")
```
