```python
!pip install pyforces-cli
import getpass
import json
import subprocess

choice = input("Choose a pyforces option (0–9): ").strip()
answers = [choice]

if choice == "0":
    header = getpass.getpass("Paste the HTTP-header JSON from Firefox: ").strip()
    json.loads(header)  # Check that it is valid JSON before starting the CLI
    answers.append(header)

result = subprocess.run(
    ["pyforces", "config"],
    input="\n".join(answers) + "\n",
    text=True,
    capture_output=True,
)

print(result.stdout, end="")
if result.stderr:
    print(result.stderr, end="")
```
